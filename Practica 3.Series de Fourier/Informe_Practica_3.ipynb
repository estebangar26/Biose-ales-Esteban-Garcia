**INTRODUCCIÓN**

La representación de señales periódicas mediante las Series de Fourier permite descomponer funciones complejas en una suma ponderada de componentes senoidales o exponenciales complejas armónicamente relacionadas [1]. El presente informe evalúa el impacto del truncamiento de la serie (uso de un número finito de armónicos $K$) en la reconstrucción de señales discontinuas y continuas, valiendose del algoritmo de FFT, provisto por la librería NumPy [3] y graficados con Matplotlib [4]. Asimismo, se aborda la caracterización espectral de una señal electrocardiográfica (ECG) limpia, vinculando la distribución de sus componentes de frecuencia con la fisiología del ciclo cardíaco.  

**ANALISIS DE LOS RESULTADOS****

Reconstrucción Progresiva y Fenómeno de Gibbs

El truncamiento de la Serie de Fourier ($K \in \{1, 3, 5, 20, 50\}$) expone diferencias fundamentales en la velocidad de convergencia según la continuidad de la señal:   

*Señal Cuadrada (Discontinua): Sus coeficientes de Fourier decaen lentamente.   Al incrementar $K$, la aproximación mejora en las regiones planas, pero persisten oscilaciones (rizado) y un sobrepico característico concentrado en las discontinuidades de salto.   Este comportamiento corresponde al Fenómeno de Gibbs. A pesar de aumentar $K \to \infty$, la magnitud del primer sobrepico no disminuye a cero, sino que se estabiliza de forma asintótica en un $8.95\%$ de la altura de la discontinuidad (Constante de Wilbraham-Gibbs) [1]. Lo que cambia al aumentar $K$ es la frecuencia de las oscilaciones, las cuales se comprimen más cerca del punto de salto.   

*Señal Triangular (Continua): Sus coeficientes de Fourier decaen de forma sustancialmente más rápida.   Debido a la continuidad en el dominio del tiempo, la reconstructed converge rápidamente sin presentar el fenómeno de Gibbs. Con tan solo $K = 5$ armónicos, la reconstrucción alcanza una fidelidad morfológica muy alta con respecto a la señal original.   


Análisis Espectral e Interpretación Fisiológica de la Señal ECGLa señal ECG es una bioseñal cuasi-periódica cuya composición en frecuencia refleja eventos electrofisiológicos específicos del corazón [2]: 

| Componente ECG | Rango Frecuencial | Correlación Fisiológica |
| :--- | :--- | :--- |
| **Onda P** | $0.5 \text{ Hz} - 10 \text{ Hz}$ | Despolarización auricular (lenta) |
| **Complejo QRS** | $8 \text{ Hz} - 40 \text{ Hz}$ | Despolarización ventricular (rápida) |
| **Onda T** | $0.5 \text{ Hz} - 7 \text{ Hz}$ | Repolarización ventricular (lenta) |

* **Complejo QRS:** Constituye el evento eléctrico de mayor amplitud y pendiente más pronunciada. Dado que las variaciones rápidas en el tiempo equivalen a componentes de alta frecuencia en el dominio espectral, el complejo QRS aporta la mayor energía en el rango medio del espectro ($8 \text{ Hz} \text{ a } 40 \text{ Hz}$).


* **Ondas P y T:** Representan cambios de voltaje suaves y de baja pendiente en la superficie corporal. Por ende, concentran su energía en las bajas frecuencias del espectro ($< 10 \text{ Hz}$).

**CONCLUSIONES:**

*Efecto de la suavidad en la convergencia de Fourier: Las señales con discontinuidades de salto (como la onda cuadrada) requieren un espectro más amplio para ser reconstruidas y sufren la limitación matemática del fenómeno de Gibbs. En contraposición, las señales continuas (como la onda triangular) concentran su energía en los primeros armónicos, logrando una reconstrucción casi exacta con pocos componentes.   

*Relevancia del fenómeno de Gibbs en procesamiento de bioseñales: En el filtrado de señales biomédicas mediante truncamiento en frecuencia (filtros Pasa-Bajo ideales), el fenómeno de Gibbs puede introducir artefactos de oscilación artificiales (efecto de ringing) en eventos de alta pendiente como el complejo QRS del ECG o transitorios EMG.   

*Utilidad de la Serie de Fourier en bioseñales: La caracterización frecuencial del ECG demuestra que la mayor parte de la información diagnóstica está contenida por debajo de los $40 \text{ Hz}$, lo cual permite diseñar sistemas eficientes de compresión, filtrado de ruido de red eléctrica ($60 \text{ Hz}$) y atenuación de artefactos de línea base.
