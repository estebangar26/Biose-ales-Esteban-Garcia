Series de Fourier: magnitud y fase.







**INTRODUCCIÓN**

En el análisis y representación de señales biomédicas y de ingeniería, las series de Fourier constituyen una herramienta matemática fundamental [1]. Estas permiten descomponer cualquier señal periódica en una suma ponderada de componentes sinusoidales de distintas frecuencias (armónicos), facilitando su estudio en el dominio espectral [1]. Sin embargo, al aproximar señales del mundo real mediante un número finito de términos de una serie de Fourier, surge un comportamiento matemático particular conocido como el fenómeno de Gibbs [1].

Este fenómeno se manifiesta como oscilaciones espurias o sobreimpulsos inevitables alrededor de las discontinuidades o cambios abruptos de pendiente (esquinas y transiciones agudas) [1]. Debido a la convergencia no uniforme de las series de Fourier cerca de estos puntos de salto, las crestas y valles de las ondas reconstruidas tienden a sobrepasar el valor real de la señal en un porcentaje constante (aproximadamente un $9\%$ del tamaño del salto), independientemente de cuántos armónicos se añadan a la sumatoria [1, 2].Comprender la naturaleza de las series de Fourier y el impacto del fenómeno de Gibbs resulta crucial en el filtrado y la síntesis de señales, ya que estas oscilaciones artificiales pueden interpretarse erróneamente como componentes de alta frecuencia genuinos [2]. Para ilustrar estos conceptos teóricos y prácticos, este informe implementa el análisis espectral y temporal sobre un caso de estudio real: una señal de electrocardiograma (ECG) registrada a una frecuencia de muestreo de $1024\text{ Hz}$ [2], donde las transiciones pronunciadas del complejo QRS permiten evidenciar de manera clara la interacción entre el desarrollo armónico de Fourier y los efectos de borde en la representación espectral final [2].

**ANALISIS DE LOS RESULTADOS****

Al evaluar el fénomeno de GIBBS  mediante señales sintéticas como la onda cuadrada y la triangular, se observan diferencias sustanciales: la señal cuadrada presenta discontinuidades y saltos verticales instantáneos, requiriendo una cantidad infinita de armónicos impares cuya truncación genera de manera inevitable el fenómeno de Gibbs, caracterizado por sobreimpulsos u oscilaciones espurias del orden del $9\%$ cerca de los bordes y un decaimiento espectral proporcional a $1/f$. En contraparte, la señal triangular es continua y posee una derivada suave por tramos, logrando que sus componentes armónicos decaigan de manera inversamente proporcional al cuadrado de la frecuencia ($1/f^2$), lo que atenúa drásticamente el fenómeno de Gibbs y proporciona una convergencia uniforme sin sobreimpulsos severos. 

  Al trasladar estos principios teóricos al análisis práctico del caso de estudio aplicado sobre el registro electrocardiográfico (ECG) procesado a una frecuencia de muestreo de $1024\text{ Hz}$, se consolida la relación directa entre el dominio del tiempo y el dominio de la frecuencia. En el dominio temporal, el registro permite visualizar con alta nitidez los ciclos cardíacos, destacando la morfología de la onda P, el complejo QRS y la onda T. Por su parte, la transformada rápida de Fourier (FFT) demuestra que la mayor parte de la energía útil de la señal se concentra por debajo de los $40\text{ Hz}$. Específicamente, las ondas P y T se sitúan en el rango de $0.5\text{ Hz}$ a $10\text{ Hz}$ debido a sus pendientes suaves y prolongadas, mientras que el complejo QRS —al actuar como una transición rápida y estrecha con alta pendiente— concentra sus componentes frecuenciales superiores entre $10\text{ Hz}$ y $40\text{ Hz}$. De este modo, comprender la interacción entre el desarrollo armónico de Fourier y los efectos de borde asociados al fenómeno de Gibbs resulta indispensable para el diseño de filtros digitales en instrumentación biomédica, asegurando la eliminación adecuada de ruido sin distorsionar los componentes de alta frecuencia esenciales para la interpretación clínica del complejo QRS.   

**CONCLUSIONES:**

Las series de Fourier constituyen un pilar teórico fundamental en el procesamiento de señales, ya que permiten descomponer cualquier forma de onda periódica en una suma ponderada de componentes sinusoidales distribuidas en el dominio de la frecuencia. No obstante, al intentar representar señales del mundo real mediante una aproximación con un número finito de armónicos, se manifiesta de forma inevitable el fenómeno de Gibbs. Este comportamiento matemático demuestra que, en las proximidades de discontinuidades o cambios abruptos de pendiente, la convergencia no uniforme de la serie genera oscilaciones espurias y sobreimpulsos de aproximadamente el $9\%$ del tamaño del salto. Dicho efecto evidencia que la reconstrucción perfecta de una señal con transiciones discontinuas es inalcanzable mediante sumatorias finitas, introduciendo artefactos que no existen en la señal original.

En el contexto específico de las bioseñales, como el electrocardiograma (ECG), las implicaciones del fenómeno de Fourier y del efecto de Gibbs adquieren una relevancia clínica crítica. Las bioseñales combinan componentes de baja frecuencia y pendientes suaves —asociadas a la onda P y la onda T— con transiciones agudas y de alta pendiente características del complejo QRS. Al aplicar filtros digitales, truncar armónicos o realizar transformadas espectrales, la presencia del fenómeno de Gibbs puede distorsionar de manera artificial la morfología de la señal, provocando sobreimpulsos que imiten patologías o enmascaren componentes reales. Por lo tanto, comprender y mitigar estos efectos mediante ventanas de ponderación adecuadas y un diseño riguroso del ancho de banda es indispensable para garantizar que el análisis espectral preserve la fidelidad fisiológica y evite diagnósticos erróneos en la instrumentación médica moderna.
