```python
import numpy as np
import matplotlib.pyplot as plt

#%%
t = np.arange(-2,2,0.01)
xt = np.sin(6*np.pi*t + (np.pi/2)) + 0.5*np.cos(4*np.pi*t + (np.pi/4)) + 2*np.sin(8*np.pi*t + (np.pi/8)) + 5

plt.plot(t,xt)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#%% SENAL CUADRADA
armonicos = np.zeros((200))
t = np.arange(-10,10,0.1)

for k in range(-1000,1000,1):
    if k == 0:

        armonicos = armonicos + 0.5
    else:
        #print(k)
        #print(np.real((1/(k*np.pi))*np.sin(k*(np.pi/2))))
        armonico = np.real((1/(k*np.pi))*np.sin(k*(np.pi/2))*np.exp((1j)*k*(np.pi/2)*t))
        armonicos = armonicos + armonico
    plt.plot(t,armonico)
plt.show()
plt.plot(t,armonicos)
plt.show()

#como se graficaría la magnitud y la fase?
```

```python
k = np.arange(-20,20,1)
Xk = (1/(k*np.pi))*np.sin(k*(np.pi/2))
Xk[20] = 0.5

plt.subplot(3,1,1)
plt.stem(Xk)
plt.subplot(3,1,2)
plt.stem(np.abs(Xk))
plt.subplot(3,1,3)
plt.stem(np.angle(Xk))
```
